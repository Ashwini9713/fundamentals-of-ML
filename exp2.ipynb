# Candidate-Elimination Algorithm
This notebook demonstrates the Candidate-Elimination algorithm using the `enjoysport.xlsx` dataset.

```python
import pandas as pd
import numpy as np

# Load the dataset from the Excel file
df = pd.read_excel('/content/enjoysport.xlsx')
display(df)
```

```python
def learn(concepts, target):
    """
    Candidate-Elimination Algorithm
    """
    # Initialize specific hypothesis with the first positive example
    specific_h = concepts[0].copy()
    print(f"Initialization of specific_h:\n{specific_h}\n")

    # Initialize general hypothesis with most general terms
    general_h = [["?" for _ in range(len(specific_h))] for _ in range(len(specific_h))]
    print(f"Initialization of general_h:\n{general_h}\n")

    for i, h in enumerate(concepts):
        # If positive instance
        if target[i] == "yes" or target[i] == True or target[i] == 1:
            for x in range(len(specific_h)):
                # Change values in S if they do not match with the current positive instance
                if h[x] != specific_h[x]:
                    specific_h[x] = '?'
                    general_h[x][x] = '?'

        # If negative instance
        else:
            for x in range(len(specific_h)):
                # For each attribute, if it doesn't match the specific hypothesis,
                # specialize the general hypothesis
                if h[x] != specific_h[x]:
                    general_h[x][x] = specific_h[x]
                else:
                    general_h[x][x] = '?'

        print(f"Steps of Candidate Elimination Algorithm (Instance {i+1}):")
        print(f"Specific Hypothesis S_{i+1}: {specific_h}")
        print(f"General Hypothesis G_{i+1}: {general_h}\n")

    # Filter out empty or duplicate general hypotheses
    indices = [i for i, val in enumerate(general_h) if val == ['?'] * len(specific_h)]
    for i in sorted(indices, reverse=True):
        general_h.remove(['?'] * len(specific_h))

    return specific_h, general_h

# Prepare features (concepts) and labels (target)
concepts = np.array(df.iloc[:, :-1])
target = np.array(df.iloc[:, -1])

s_final, g_final = learn(concepts, target)

print("--------------------------------------------------")
print("Final Specific Hypothesis:", s_final)
print("Final General Hypothesis:", g_final)
```
